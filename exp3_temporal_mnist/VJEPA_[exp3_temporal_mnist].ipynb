```python
"""
Experiment 3: Temporal MNIST with High-Variance Random-Image Distractors
=========================================================================
Addresses:
  - Reviewer dcLu Q1: "Have you experimented with real-world image, video,
    or other spatiotemporal data?"
  - Reviewer ZQ8k W2: "it still seems feasible to evaluate the framework on
    a simple image dataset such as MNIST"
  - Promised in rebuttal: "Temporal MNIST with high-variance random-image
    distractors, as a visual analogue of our toy setting, measuring whether
    VJEPA can extract digit-sequence dynamics while ignoring unrelated noise frames."

Design (visual analogue of the Noisy TV experiment):
  ┌─────────────────────────────────────────────────────────┐
  │  Signal:     slowly-cycling MNIST digit sequences       │
  │              digit class rotates 0→1→2→…→9→0 every K=8 │
  │              steps; each frame = random sample from     │
  │              current class (28×28, flattened to 784)    │
  │                                                         │
  │  Distractor: independently sampled random MNIST images  │
  │              (any class, high-variance "Noisy TV")      │
  │                                                         │
  │  Observation: x_t = signal_t + σ · distractor_t        │
  │               ∈ ℝ^784, normalized to [0,1]              │
  └─────────────────────────────────────────────────────────┘

Latent space: DIM_Z = 16 (enough to represent digit dynamics).
Encoder:      2-layer MLP, 784 → 256 → 16  (Tanh activations).
Evaluation:   linear probe on z_pred → digit class of NEXT frame.
              Metric: Top-1 classification accuracy (analogous to Signal R²).

Produces:
  - fig_temporal_mnist.pdf   (accuracy vs. noise scale, 3 models)
  - fig_temporal_mnist_samples.pdf  (qualitative: signal vs. distractor images)

Runtime estimate: ~25–40 min on Colab CPU  (3 scales × 3 models × 3000 epochs)
"""

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from copy import deepcopy
import time

# ── Try to import torchvision; fall back with instructions ────────────────────
try:
    import torchvision
    import torchvision.transforms as T
    HAS_TV = True
except ImportError:
    HAS_TV = False
    print("torchvision not found.  Run:  !pip install torchvision")

# ── Hyperparameters ────────────────────────────────────────────────────────────
SEED        = 111
SCALES      = [0.0, 1.0, 2.0, 3.0, 4.0]   # distractor multiplier (pixels ∈[0,1])
N_SEQ       = 4000                          # total timesteps in training sequence
N_SEQ_TEST  = 1000
CLASS_REPEAT = 8                            # how many steps each digit class lasts
DIM_X       = 784                           # 28×28 flattened
DIM_Z       = 16
HIDDEN      = 256
N_EPOCHS    = 3000
LR          = 5e-4
BETA        = 0.01
GAMMA       = 0.10
EMA_TAU     = 0.99

torch.manual_seed(SEED); np.random.seed(SEED)


# ── Data: Temporal MNIST Sequences ────────────────────────────────────────────
def load_mnist():
    """Download MNIST and return (images, labels) tensors, values in [0,1]."""
    assert HAS_TV, "torchvision required"
    ds_tr = torchvision.datasets.MNIST('/tmp/mnist', train=True,  download=True,
                                        transform=T.ToTensor())
    ds_te = torchvision.datasets.MNIST('/tmp/mnist', train=False, download=True,
                                        transform=T.ToTensor())
    imgs_tr = ds_tr.data.float() / 255.0        # [60000, 28, 28]
    lbs_tr  = ds_tr.targets                     # [60000]
    imgs_te = ds_te.data.float() / 255.0        # [10000, 28, 28]
    lbs_te  = ds_te.targets
    # Flatten
    return (imgs_tr.view(-1, DIM_X), lbs_tr,
            imgs_te.view(-1, DIM_X), lbs_te)


def build_sequence(imgs, labels, n_steps, repeat=CLASS_REPEAT, rng=None):
    """
    Build a temporal MNIST signal sequence.

    Returns:
      x_sig  : [n_steps, 784]   – one sample from current digit class per step
      y_sig  : [n_steps]        – digit class at each step (0–9)
    """
    if rng is None: rng = np.random.RandomState(SEED)
    # Pre-group images by class for fast lookup
    by_class = {c: imgs[labels == c] for c in range(10)}
    x_sig, y_sig = [], []
    n_classes = 10
    for step in range(n_steps):
        cls = (step // repeat) % n_classes
        pool = by_class[cls]
        idx = rng.randint(0, len(pool))
        x_sig.append(pool[idx])
        y_sig.append(cls)
    return torch.stack(x_sig), torch.tensor(y_sig, dtype=torch.long)


def build_distractor_sequence(imgs, n_steps, rng=None):
    """Sample n_steps random MNIST images as distractors (no temporal structure)."""
    if rng is None: rng = np.random.RandomState(SEED + 1)
    idxs = rng.randint(0, len(imgs), size=n_steps)
    return imgs[idxs]        # [n_steps, 784]


def make_dataset(imgs_src, labels_src, n_steps, scale, rng_sig=None, rng_dist=None):
    """
    Combine signal + distractor into observations.

    Returns: x_t, x_{t+1}, class_t, class_{t+1}  (temporal pairs)
    """
    x_sig, y_sig = build_sequence(imgs_src, labels_src, n_steps+1,
                                   rng=rng_sig)
    x_dist = build_distractor_sequence(imgs_src, n_steps+1, rng=rng_dist)

    # x_t = signal_t + σ * distractor_t,  clipped to [0,1] for interpretability
    x_obs = torch.clamp(x_sig + scale * x_dist, 0.0, 1.0)

    # Temporal pairs
    xt  = x_obs[:-1]          # [n_steps, 784]
    xn  = x_obs[1:]           # [n_steps, 784]
    yt  = y_sig[:-1]          # class at t
    yn  = y_sig[1:]           # class at t+1  ← what predictive models should learn

    return xt, xn, yt, yn


def show_samples(x_sig, x_dist, x_obs, scale, n=8):
    """Visualize signal, distractor, and combined observation frames."""
    fig, axes = plt.subplots(3, n, figsize=(n*1.5, 4.5))
    for i in range(n):
        axes[0,i].imshow(x_sig[i].view(28,28).numpy(), cmap='gray', vmin=0, vmax=1)
        axes[1,i].imshow(x_dist[i].view(28,28).numpy(), cmap='gray', vmin=0, vmax=1)
        axes[2,i].imshow(x_obs[i].view(28,28).numpy(), cmap='gray', vmin=0, vmax=1)
        for ax in axes[:,i]: ax.axis('off')
    axes[0,0].set_ylabel('Signal', fontsize=10)
    axes[1,0].set_ylabel(f'Distractor\n(σ={scale})', fontsize=10)
    axes[2,0].set_ylabel('Observation', fontsize=10)
    plt.suptitle('Temporal MNIST: Signal + Noisy TV Distractor', fontsize=12)
    plt.tight_layout()
    plt.savefig('fig_temporal_mnist_samples.pdf', bbox_inches='tight')
    plt.show()


# ── Models (MLP, image-capable) ───────────────────────────────────────────────
def enc_net():
    return nn.Sequential(nn.Linear(DIM_X, HIDDEN), nn.Tanh(),
                         nn.Linear(HIDDEN, DIM_Z))

def pred_net():
    return nn.Sequential(nn.Linear(DIM_Z, HIDDEN//2), nn.Tanh(),
                         nn.Linear(HIDDEN//2, DIM_Z))

class MNISTJEPA(nn.Module):
    def __init__(self):
        super().__init__()
        self.enc  = enc_net()
        self.pred = pred_net()
        self.tgt  = deepcopy(self.enc)
        for p in self.tgt.parameters(): p.requires_grad_(False)

    def forward(self, xt, xn):
        zp = self.pred(self.enc(xt))
        with torch.no_grad(): zt = self.tgt(xn)
        return zp, zt

    def ema(self):
        for p, tp in zip(self.enc.parameters(), self.tgt.parameters()):
            tp.data.mul_(EMA_TAU).add_(p.data, alpha=1-EMA_TAU)

    def latent(self, x): return self.pred(self.enc(x))


class MNISTVJEPA(nn.Module):
    def __init__(self):
        super().__init__()
        self.enc         = enc_net()
        self.shared_pred = nn.Sequential(nn.Linear(DIM_Z, HIDDEN//2), nn.Tanh())
        self.pred_mu     = nn.Linear(HIDDEN//2, DIM_Z)
        self.pred_logvar = nn.Linear(HIDDEN//2, DIM_Z)
        self.tgt_mu      = deepcopy(self.enc)
        self.tgt_logvar  = nn.Sequential(nn.Linear(DIM_X, HIDDEN//2), nn.Tanh(),
                                          nn.Linear(HIDDEN//2, DIM_Z))
        for p in self.tgt_mu.parameters():     p.requires_grad_(False)
        for p in self.tgt_logvar.parameters(): p.requires_grad_(False)

    def forward(self, xt, xn):
        h = self.shared_pred(self.enc(xt))
        p_mu, p_lv = self.pred_mu(h), self.pred_logvar(h)
        with torch.no_grad():
            t_mu, t_lv = self.tgt_mu(xn), self.tgt_logvar(xn)
        z_s = t_mu + (0.5*t_lv).exp() * torch.randn_like(t_mu)
        return z_s, (p_mu, p_lv), (t_mu, t_lv)

    def ema(self):
        for p, tp in zip(self.enc.parameters(), self.tgt_mu.parameters()):
            tp.data.mul_(EMA_TAU).add_(p.data, alpha=1-EMA_TAU)

    def latent(self, x):
        h = self.shared_pred(self.enc(x))
        return self.pred_mu(h)

    def pred_variance(self, x):
        with torch.no_grad():
            h  = self.shared_pred(self.enc(x))
            lv = self.pred_logvar(h)
        return lv.exp().mean(0)


class MNISTBJEPA(nn.Module):
    def __init__(self):
        super().__init__()
        self.enc         = enc_net()
        self.shared_pred = nn.Sequential(nn.Linear(DIM_Z, HIDDEN//2), nn.Tanh())
        self.pred_mu     = nn.Linear(HIDDEN//2, DIM_Z)
        self.pred_logvar = nn.Linear(HIDDEN//2, DIM_Z)
        self.tgt_mu      = deepcopy(self.enc)
        self.tgt_logvar  = nn.Sequential(nn.Linear(DIM_X, HIDDEN//2), nn.Tanh(),
                                          nn.Linear(HIDDEN//2, DIM_Z))
        self.prior_mu    = nn.Parameter(torch.zeros(DIM_Z))
        self.prior_lv    = nn.Parameter(torch.zeros(DIM_Z))
        for p in self.tgt_mu.parameters():     p.requires_grad_(False)
        for p in self.tgt_logvar.parameters(): p.requires_grad_(False)

    @staticmethod
    def poe(mu1, lv1, mu2, lv2):
        pr1, pr2 = (-lv1).exp(), (-lv2).exp()
        pr = pr1 + pr2
        return (mu1*pr1 + mu2*pr2)/pr, -pr.log()

    def forward(self, xt, xn):
        h = self.shared_pred(self.enc(xt))
        d_mu, d_lv = self.pred_mu(h), self.pred_logvar(h)
        B = xt.size(0)
        pr_mu = self.prior_mu.unsqueeze(0).expand(B, -1)
        pr_lv = self.prior_lv.unsqueeze(0).expand(B, -1)
        with torch.no_grad():
            t_mu, t_lv = self.tgt_mu(xn), self.tgt_logvar(xn)
        z_s = t_mu + (0.5*t_lv).exp() * torch.randn_like(t_mu)
        return z_s, (d_mu, d_lv), (pr_mu, pr_lv), (t_mu, t_lv)

    def ema(self):
        for p, tp in zip(self.enc.parameters(), self.tgt_mu.parameters()):
            tp.data.mul_(EMA_TAU).add_(p.data, alpha=1-EMA_TAU)

    def latent(self, x):
        h = self.shared_pred(self.enc(x))
        d_mu, d_lv = self.pred_mu(h), self.pred_logvar(h)
        B = x.size(0)
        pr_mu = self.prior_mu.unsqueeze(0).expand(B, -1)
        pr_lv = self.prior_lv.unsqueeze(0).expand(B, -1)
        mu, _ = self.poe(d_mu, d_lv, pr_mu, pr_lv)
        return mu


# ── Losses ─────────────────────────────────────────────────────────────────────
def vicreg(x, y, sim=25., std=25., cov=1.):
    loss = sim * nn.functional.mse_loss(x, y)
    loss += std * (torch.relu(1-(x.var(0)+1e-4).sqrt()).mean() +
                   torch.relu(1-(y.var(0)+1e-4).sqrt()).mean())
    x, y = x-x.mean(0), y-y.mean(0); N, D = x.shape
    cx = x.T @ x / (N-1)
    loss += cov * (cx.pow(2).sum() - cx.diag().pow(2).sum()) / D
    return loss

def vjepa_loss(z_s, p_mu, p_lv, t_mu, t_lv):
    pv  = p_lv.exp()
    nll = 0.5*(pv.log()+(z_s-p_mu).pow(2)/pv).sum(-1).mean()
    kl  = -0.5*(1+t_lv-t_mu.pow(2)-t_lv.exp()).sum(-1).mean()
    return nll + BETA*kl

def bjepa_loss(z_s, d_mu, d_lv, pr_mu, pr_lv, t_mu, t_lv):
    base = vjepa_loss(z_s, d_mu, d_lv, t_mu, t_lv)
    vr   = (d_lv-pr_lv).exp()
    kl   = 0.5*(vr+(pr_mu-d_mu).pow(2)/pr_lv.exp()-1-(d_lv-pr_lv)).sum(-1).mean()
    return base + GAMMA*kl


# ── Training ───────────────────────────────────────────────────────────────────
BATCH = 256

def train_model(mdl, name, xt, xn):
    opt = optim.Adam(mdl.parameters(), lr=LR)
    N   = xt.size(0)
    for epoch in range(N_EPOCHS):
        perm  = torch.randperm(N)
        xt_sh = xt[perm]; xn_sh = xn[perm]
        for i in range(0, N, BATCH):
            xb, xnb = xt_sh[i:i+BATCH], xn_sh[i:i+BATCH]
            opt.zero_grad()
            if name == 'JEPA':
                zp, zt = mdl(xb, xnb); loss = vicreg(zp, zt)
            elif name == 'VJEPA':
                z_s,(pm,pl),(tm,tl) = mdl(xb, xnb)
                loss = vjepa_loss(z_s,pm,pl,tm,tl)
            elif name == 'BJEPA':
                z_s,(dm,dl),(pm,pl),(tm,tl) = mdl(xb, xnb)
                loss = bjepa_loss(z_s,dm,dl,pm,pl,tm,tl)
            loss.backward(); opt.step()
        # EMA once per epoch — consistent with exp1/exp2 (full-batch τ=0.99)
        if name in ('JEPA', 'VJEPA', 'BJEPA'):
            mdl.ema()


# ── Evaluation: digit classification accuracy ─────────────────────────────────
def eval_acc(mdl, xt_tr, yn_tr, xt_te, yn_te):
    """
    Fit a logistic regression probe on z_pred to predict the NEXT digit class.
    Returns (train_acc, test_acc).
    """
    with torch.no_grad():
        z_tr = mdl.latent(xt_tr).numpy()
        z_te = mdl.latent(xt_te).numpy()
    clf = LogisticRegression(max_iter=500, C=1.0, solver='lbfgs', multi_class='auto')
    clf.fit(z_tr, yn_tr.numpy())
    acc_tr = accuracy_score(yn_tr.numpy(), clf.predict(z_tr))
    acc_te = accuracy_score(yn_te.numpy(), clf.predict(z_te))
    return acc_tr, acc_te


# ── Main ───────────────────────────────────────────────────────────────────────
def run():
    imgs_tr, lbs_tr, imgs_te, lbs_te = load_mnist()
    print(f"MNIST loaded: train={len(imgs_tr)}, test={len(imgs_te)}")

    # Show qualitative samples at scale=3
    rng_s = np.random.RandomState(SEED); rng_d = np.random.RandomState(SEED+1)
    x_sig_vis, _ = build_sequence(imgs_tr, lbs_tr, 8, rng=rng_s)
    x_dis_vis    = build_distractor_sequence(imgs_tr, 8, rng=rng_d)
    x_obs_vis    = torch.clamp(x_sig_vis + 3.0 * x_dis_vis, 0, 1)
    show_samples(x_sig_vis, x_dis_vis, x_obs_vis, scale=3.0)

    model_names = ['JEPA', 'VJEPA', 'BJEPA']
    CLASSES     = {n: {'acc_tr':[], 'acc_te':[]} for n in model_names}

    hdr = f"{'Scale':>6} | {'Model':>6} | {'AccTr':>7} | {'AccTe':>7} | Time"
    print(hdr); print("-"*len(hdr))

    for scale in SCALES:
        rng_s_tr  = np.random.RandomState(SEED)
        rng_d_tr  = np.random.RandomState(SEED+10)
        rng_s_te  = np.random.RandomState(SEED+2)
        rng_d_te  = np.random.RandomState(SEED+3)

        xt_tr, xn_tr, yt_tr, yn_tr = make_dataset(
            imgs_tr, lbs_tr, N_SEQ,      scale, rng_s_tr, rng_d_tr)
        xt_te, xn_te, yt_te, yn_te = make_dataset(
            imgs_te, lbs_te, N_SEQ_TEST, scale, rng_s_te, rng_d_te)

        # Normalize using training statistics
        mu_n, std_n = xt_tr.mean(0), xt_tr.std(0)+1e-6
        xt_tr_n  = (xt_tr  - mu_n) / std_n
        xn_tr_n  = (xn_tr  - mu_n) / std_n
        xt_te_n  = (xt_te  - mu_n) / std_n

        for name in model_names:
            t0  = time.time()
            mdl = {'JEPA': MNISTJEPA, 'VJEPA': MNISTVJEPA, 'BJEPA': MNISTBJEPA}[name]()
            train_model(mdl, name, xt_tr_n, xn_tr_n)
            acc_tr, acc_te = eval_acc(mdl, xt_tr_n, yn_tr, xt_te_n, yn_te)
            CLASSES[name]['acc_tr'].append(acc_tr)
            CLASSES[name]['acc_te'].append(acc_te)
            elapsed = time.time()-t0
            print(f"{scale:>6.1f} | {name:>6} | {acc_tr:>7.3f} | {acc_te:>7.3f} | {elapsed:.0f}s")

        print("-"*len(hdr))

    # ── Plot: Accuracy vs. Noise Scale ────────────────────────────────────────
    colors = {'JEPA':'#17becf', 'VJEPA':'#1f77b4', 'BJEPA':'#9467bd'}
    markers = {'JEPA':'v', 'VJEPA':'s', 'BJEPA':'D'}
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))

    for split, ax in zip(['acc_tr', 'acc_te'], axes):
        for name in model_names:
            ax.plot(SCALES, CLASSES[name][split], label=name,
                    color=colors[name], marker=markers[name], linewidth=2.5)
        ax.axhline(0.1, color='gray', linestyle=':', label='Chance (10%)')
        ax.set_title(f"{'Train' if 'tr' in split else 'Test'} Digit Prediction Accuracy",
                     fontsize=13)
        ax.set_xlabel('Distractor noise scale σ', fontsize=12)
        ax.set_ylabel('Next-frame digit accuracy', fontsize=12)
        ax.legend(fontsize=11); ax.grid(alpha=0.3)
        ax.set_ylim(0, 1.05)

    plt.suptitle('Experiment 3: Temporal MNIST with Noisy-TV Distractors\n'
                 'VJEPA/BJEPA maintain digit accuracy as distractor noise increases',
                 fontsize=12)
    plt.tight_layout()
    plt.savefig('fig_temporal_mnist.pdf', bbox_inches='tight')
    plt.show()
    print("Saved: fig_temporal_mnist.pdf")

    # ── Per-dim variance at scale=3 ───────────────────────────────────────────
    print("\nGenerating per-dimension variance figure for VJEPA …")
    scale_for_var = 3.0
    rng_sv = np.random.RandomState(SEED)
    rng_dv = np.random.RandomState(SEED+10)
    xt_v, xn_v, _, _ = make_dataset(imgs_tr, lbs_tr, N_SEQ, scale_for_var, rng_sv, rng_dv)
    mu_v, std_v = xt_v.mean(0), xt_v.std(0)+1e-6
    xt_v_n = (xt_v - mu_v) / std_v

    mdl_v = MNISTVJEPA()
    train_model(mdl_v, 'VJEPA', (xt_v - mu_v)/std_v, (xn_v - mu_v)/std_v)
    var_dims = mdl_v.pred_variance(xt_v_n).numpy()

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.bar(range(DIM_Z), np.sort(var_dims), color='steelblue', edgecolor='k')
    ax.axhline(scale_for_var**2, color='crimson', linestyle='--',
               label=f'Prior σ²={scale_for_var**2:.0f} (noise level)')
    ax.set_xlabel('Latent dimension (sorted by variance)', fontsize=12)
    ax.set_ylabel('Mean predictive variance', fontsize=12)
    ax.set_title('VJEPA: Per-Dimension Predictive Variance (Temporal MNIST, σ=3)\n'
                 'Dimensions encoding noise → variance ≈ prior σ²I  (Prop. 5.2)',
                 fontsize=11)
    ax.legend(fontsize=11); ax.grid(alpha=0.3, axis='y')
    plt.tight_layout()
    plt.savefig('fig_temporal_mnist_variance.pdf', bbox_inches='tight')
    plt.show()
    print("Saved: fig_temporal_mnist_variance.pdf")

    return CLASSES


if __name__ == '__main__':
    if not HAS_TV:
        print("Install torchvision first:  !pip install torchvision")
    else:
        run()
```
